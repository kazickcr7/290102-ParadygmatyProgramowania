```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import time

# ==============================================================================
# 1. FUNKCJE BAZOWE (z materiałów od prowadzącego / interpolation.py)
# ==============================================================================
def Π(x, l=-0.5, r=0.5):
    # Funkcja prostokątna (najbliższy sąsiad) - zwraca 1 w przedziale, 0 poza nim
    return (x >= l) * (x < r)

def Λ(x):
    # Funkcja trójkątna (interpolacja liniowa)
    return (1 - np.abs(x)) * Π(x, -1, 1)

def φ(x):
    # Sześcienna funkcja interpolująca Keysa (dla alfa = -0.5)
    x, α = np.abs(x), -0.5
    return (((α + 2) * x**3 - (α + 3) * x**2 + 1) * Π(x, 0, 1)
            + (α * x**3 - 5 * α * x**2 + 8 * α * x - 4 * α) * Π(x, 0, 2) * (1 - Π(x, 0, 1)))

def Φ(x, f, ϕ):
    # Interpolacja 1D dla pojedynczego punktu x
    n = np.arange(len(f))
    ψ = ϕ(x - n)
    return ψ @ f

def interpolate(fn, N, ϕ):
    # Przeliczenie całego wektora próbek na nową długość N wybraną metodą ϕ
    X = np.linspace(0, len(fn) - 1, N)
    return np.array([Φ(x, fn, ϕ) for x in X])

# ==============================================================================
# 2. ALGORYTMY SKALOWANIA (C1) I OBROTU (C2)
# ==============================================================================
def skaluj_2d(img, skala, metoda):
    """
    Skalowanie obrazu 2D wykorzystujące separowalność (najpierw wiersze, potem kolumny).
    Działa dla obrazów w kolorze (RGB).
    """
    M, N, C = img.shape
    nowe_M = int(round(M * skala))
    nowe_N = int(round(N * skala))

    # Tablice pomocnicze na wynik pośredni (po wierszach) i końcowy
    posredni = np.zeros((M, nowe_N, C))
    wynik = np.zeros((nowe_M, nowe_N, C))

    for c in range(C):  # Lecimy osobno po każdym kanale R, G, B
        # Krok 1: interpolacja wierszy obrazu
        for m in range(M):
            posredni[m, :, c] = interpolate(img[m, :, c], nowe_N, metoda)
        # Krok 2: interpolacja kolumn z macierzy pośredniej
        for n in range(nowe_N):
            wynik[:, n, c] = interpolate(posredni[:, n, c], nowe_M, metoda)

    # Przycięcie wartości do bezpiecznego zakresu [0.0, 1.0]
    return np.clip(wynik, 0.0, 1.0)

def obroc_2d(img, kat_stopnie, metoda):
    """
    Obrót 2D wokół środka obrazu.
    Używamy mapowania wstecznego (backward mapping) z kątem -alpha,
    żeby w obrazie wynikowym nie powchodziły nam czarne dziury/puste piksele.
    """
    M, N, C = img.shape
    wynik = np.zeros_like(img)

    # Wyznaczamy środek obrazu zgodnie z instrukcją
    x0, y0 = (N - 1) / 2.0, (M - 1) / 2.0

    # Przeliczamy kąt na radiany i bierzemy minus alpha do mapowania wstecznego
    alpha = -np.deg2rad(kat_stopnie)
    cos_a, sin_a = np.cos(alpha), np.sin(alpha)

    # Tworzymy siatkę współrzędnych dla docelowego obrazu
    y_idx, x_idx = np.indices((M, N))
    dx = x_idx - x0
    dy = y_idx - y0

    # Wzór na obrót punktów z uwzględnieniem środka kadru
    x_stare = cos_a * dx + sin_a * dy + x0
    y_stare = -sin_a * dx + cos_a * dy + y0

    # Odrzucamy to, co po obrocie wyleci poza oryginalne wymiary obrazka
    maska = (x_stare >= 0) & (x_stare <= N - 1) & (y_stare >= 0) & (y_stare <= M - 1)
    y_valid, x_valid = np.where(maska)
    xs = x_stare[maska]
    ys = y_stare[maska]

    n_idx = np.arange(N)
    m_idx = np.arange(M)

    # Szybkie liczenie wag dla wszystkich ważnych punktów na raz
    wagi_x = metoda(xs[:, None] - n_idx[None, :])  # wagi w poziomie
    wagi_y = metoda(ys[:, None] - m_idx[None, :])  # wagi w pionie

    for c in range(C):
        # Dwuwymiarowa interpolacja ważona dla pikseli, które zmieściły się w masce
        wartosci = np.einsum('pm,mn,pn->p', wagi_y, img[:, :, c], wagi_x)
        wynik[y_valid, x_valid, c] = wartosci

    return np.clip(wynik, 0.0, 1.0)

def oblicz_bledy(oryginal, wynik):
    """ Liczy MSE oraz PSNR do naszej tabelki w sprawozdaniu. """
    # Zabezpieczenie, gdyby przez zaokrąglenia wymiary różniły się o 1 piksel
    M = min(oryginal.shape[0], wynik.shape[0])
    N = min(oryginal.shape[1], wynik.shape[1])
    oryg = oryginal[:M, :N, :]
    wyn = wynik[:M, :N, :]

    mse = np.mean((oryg - wyn) ** 2)
    psnr = 10 * np.log10(1.0 / mse) if mse > 0 else float('inf')
    return mse, psnr

# ==============================================================================
# 3. GŁÓWNA PĘTLA EKSPERYMENTÓW (C3 - C10)
# ==============================================================================
NAZWA_PLIKU = 'images/Pasikonik.jpg'

# Wczytujemy obrazek i normalizujemy jego piksele do zakresu [0.0, 1.0]
img_oryg = np.asarray(Image.open(NAZWA_PLIKU), dtype=np.float64) / 255.0

# Opcjonalnie zmniejszam szerokość do 200px, żeby algorytm Keysa nie liczył się wieki na labach
pil_img = Image.open(NAZWA_PLIKU)
if pil_img.width > 200:
    nowa_wys = int(pil_img.height * (200.0 / pil_img.width))
    pil_img = pil_img.resize((200, nowa_wys))

img_oryg = np.asarray(pil_img, dtype=np.float64) / 255.0

# Zapisujemy obraz bazowy do porównań wizualnych
Image.fromarray((img_oryg * 255).astype(np.uint8)).save('0_oryginal.png')

# Lista metod do przetestowania zgodnie z instrukcją
metody = [
    ('Najblizszy_sasiad_Pi', Π),
    ('Liniowa_Lambda', Λ),
    ('Szescienna_Keysa', φ)
]

print(f"\nStartuję testy! Rozmiar obrazu testowego: {img_oryg.shape[:2]}\n")
print(f"{'Metoda':<22} | {'Test':<18} | {'MSE':<12} | {'PSNR [dB]':<10} | {'Czas [s]'}")
print("-" * 78)

for nazwa, funkcja in metody:
    # --- TEST 1 (C3-C5): Powiększamy 5 razy o 10%, a potem tyle samo pomniejszamy ---
    t0 = time.time()
    obr = img_oryg.copy()
    for _ in range(5):
        obr = skaluj_2d(obr, 1.1, funkcja)
    for _ in range(5):
        obr = skaluj_2d(obr, 1 / 1.1, funkcja)
    czas = time.time() - t0
    mse, psnr = oblicz_bledy(img_oryg, obr)
    Image.fromarray((obr * 255).astype(np.uint8)).save(f'Test1_{nazwa}.png')
    print(f"{nazwa:<22} | {'1. Powieksz/Pomn':<18} | {mse:<12.6f} | {psnr:<10.2f} | {czas:.2f}")

    # --- TEST 2 (C6-C8): Najpierw pomniejszamy 3 razy o 10%, a potem powiększamy ---
    t0 = time.time()
    obr = img_oryg.copy()
    for _ in range(3):
        obr = skaluj_2d(obr, 1 / 1.1, funkcja)
    for _ in range(3):
        obr = skaluj_2d(obr, 1.1, funkcja)
    czas = time.time() - t0
    mse, psnr = oblicz_bledy(img_oryg, obr)
    Image.fromarray((obr * 255).astype(np.uint8)).save(f'Test2_{nazwa}.png')
    print(f"{nazwa:<22} | {'2. Pomn/Powieksz':<18} | {mse:<12.6f} | {psnr:<10.2f} | {czas:.2f}")

    # --- TEST 3 (C9-C10): Kręcimy obrazem 30 razy po 12 stopni (= pełne 360 stopni) ---
    t0 = time.time()
    obr = img_oryg.copy()
    for _ in range(30):
        obr = obroc_2d(obr, 12, funkcja)
    czas = time.time() - t0
    mse, psnr = oblicz_bledy(img_oryg, obr)
    Image.fromarray((obr * 255).astype(np.uint8)).save(f'Test3_{nazwa}.png')
    print(f"{nazwa:<22} | {'3. Obrot 30x12st':<18} | {mse:<12.6f} | {psnr:<10.2f} | {czas:.2f}")
    print("-" * 78)

print("\nKoniec! Obrazki wynikowe zapisały się w folderze roboczym.")
```
